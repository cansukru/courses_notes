## Lists

### Indexing

We are going to take a look at lists in Python. A list is a sequenced collection of different objects such as integers, strings, and even other lists as well. The address of each element within a list is called an index. An index is used to access and refer to items within a list.

## List Content
Lists can contain strings, floats, and integers. We can nest other lists, and we can also nest tuples and other data structures. The same indexing conventions apply for nesting:


```python
# Sample List
["Michael Jackson", 10.1, 1982, [1, 2], ("A", 1)]
```

## List Operations
We can also perform slicing in lists. For example, if we want the last two elements, we use the following command:

```python
# Sample List
L = ["Michael Jackson", 10.1,1982,"MJ",1]
L
```

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%202/images/ListsSlice.png" width="1000">


```python
# List slicing
L[3:5]
```

```python
# Use extend to add elements to list
L = [ "Michael Jackson", 10.2]
L.extend(['pop', 10])
L
```

```python
# Use append to add elements to list
L = [ "Michael Jackson", 10.2]
L.append(['pop', 10])
L
```

```python
# Change the element based on the index

A = ["disco", 10, 1.2]
print('Before change:', A)
A[0] = 'hard rock'
print('After change:', A)
```

```python
# Delete the element based on the index
A = ["disco", 10, 1.2]
print('Before change:', A)
del(A[0])
print('After change:', A)
```

```python
# Split the string, default is by space

'hard rock'.split()
```

```python
# Split the string by comma

'A,B,C,D'.split(',')
```

<h3 id="co">Copy and Clone List</h3>


```python
# Copy (copy by reference) the list A
A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
```

When Python writes `B = A`, it doesn't create a new list. It simply binds the variable named `B` to the same list object that `A` points to. So, there's only one list in memory, but you can access it with two names (A and B).

Initially, the value of the first element in <b>B</b> is set as "hard rock". If we change the first element in <b>A</b> to <b>"banana"</b>, we get an unexpected side effect.  As <b>A</b> and <b>B</b> are referencing the same list, if we change list <b>A</b>, then list <b>B</b> also changes. If we check the first element of <b>B</b> we get "banana" instead of "hard rock":

```python
print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```

```python
A = ["hard rock", 10, 1.2]
# Clone (clone by value) the list A
B = A[:]
B
```

The expression A[:] copies all the elements of list A from beginning to end into a new list. Thus, there are now two separate list objects in memory: one A and one B. Although their contents are the same, they are independent of each other.

```python
print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```
